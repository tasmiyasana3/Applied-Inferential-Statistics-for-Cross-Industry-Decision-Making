# **PROBLEM 2**

##**Context**

**The breaking strength of gunny bags used for packaging cement is normally distributed with a mean of 5 kg per sq. centimeter and a standard deviation of 1.5 kg per sq. centimeter. The quality team of the cement company wants to know the following about the packaging material to better understand wastage or pilferage within the supply chain; Answer the questions below based on the given information.**




**Data Dictionary:**

* **This problem is of Normal distribution.**
* **Mean(mu) breaking strength = 5kg/sq.cm . --> Population mean.**
* **Standard deviation(sigma) = 1.5Kg/sq.cm  --> Population std. deviation.**
* **Since, the std. deviation is known we will use Z-test.**



##**2.1 What proportion of the gunny bags have a breaking strength of less than 3.17 kg per sq cm?**

**Calculations:**

```python
mu1 = 5  #mean breaking strength.
sigm1 = 1.5 #std deviation.
#P(X<3.17)
x1 = 3.17
z1 = (x1-mu1)/sigm1
print('Z-score is %0.2f' %(z1))

p1 = norm.cdf(3.17, loc=mu1, scale=sigm1) #using cummulative distributive fucntion to calculate probability.
print("The proportion of the gunny bags have a breaking strength of less than 3.17 kg per sq cm is", str(round(p1,4)*100)+"%")
```

**Visualisation:**

```python
#Generate x and y for normal distribution curve.
x = np.linspace(mu1-4*sigm1, mu1+4*sigm1, 1000)
y = norm.pdf(x, mu1, sigm1 )
#plot figure size.
plt.figure(figsize=(8,6))
#plot normal distribution curve.
plt.plot(x,y,label='Normal Distribution')
#fills area under the curve left of X<=3.17.
plt.fill_between(x,y, where=(x<=x1), color='green',alpha=0.5)
#Draws vertical dashed line at x1 to mark threshold.
plt.axvline(x1, color='red', linestyle='--', label=f'X={x1}')
#plot title
plt.title(f'2.1: P(X<{x1}) = {p1:.4f}')
# x-axis label
plt.xlabel('Breaking Strength(Kg/sq.cm)') # x-axis label
plt.ylabel('Probability Density') #y-axis label.
plt.legend()
plt.grid(True)
plt.show()


```

**Insights:**

*  Approximately 11.1% of the gunny bags have a breaking strength below 3.17 kg/sq. cm, which is significantly lower than the average strength (5 kg/sq. cm).
*   This indicates that more than 1 in 10 bags maybe valuable to tearing or failure during transport, having a potential risk to product safety and supply chain efficiency.

**Recommendation:**

* Investigate manufacturing variation to reduce the number of weak gunny bags.



##**2.2 What proportion of the gunny bags have a breaking strength of at least 3.6 kg per sq cm.?**

**Calculations:**

```python
x2 = 3.6
z2 = (x2-mu1)/sigm1
print('Z-score is %0.2f' %(z2))

p2 = 1- norm.cdf(x2, loc=mu1, scale=sigm1) #using cummulative distributive fucntion to calculate probability.
print("The proportion of the gunny bags have a breaking strength of at least 3.6 kg per sq cm is", str(round(p2,4)*100)+"%")
```

**Visualization:**

```python
#Generate x and y for normal distribution curve.
x = np.linspace(mu1-4*sigm1, mu1+4*sigm1, 1000)
y = norm.pdf(x, mu1, sigm1 )
#plot figure size.
plt.figure(figsize=(8,6))
#plot normal distribution curve.
plt.plot(x,y,label='Normal Distribution')
#fills area under the curve right of X>=3.6 .
plt.fill_between(x,y, where=(x>=x2), color='skyblue',alpha=0.5)
#Draws vertical dashed line at x2 to mark threshold.
plt.axvline(x2, color='green', linestyle='--', label=f'X={x2}')
#plot title
plt.title(f'2.2: P(X ≥ 3.6) ≈ {p2:.4f}')
# x-axis label
plt.xlabel('Breaking Strength(Kg/sq.cm)') # x-axis label
plt.ylabel('Probability Density') #y-axis label.
plt.legend()
plt.grid(True)
plt.show()
```

**Insights:**

*   Around 82.4% of the gunny bags have a breaking strength of 3.6 Kg/sq. cm or more.
*   This shows that the majority of the bags are strong enough to use.

**Recommendation:**

* However, since the threshold of 36 is not far below the mean (5), we need to monitor the performance and efficiency of the bags closer to the lower bound.





##**2.3 What proportion of the gunny bags have a breaking strength between 5 and 5.5 kg per sq cm.?**

**Calculations:**


```python
x3_lower = 5
x3_upper = 5.5
z3_low = (x3_lower-mu1)/sigm1
z3_upper = (x3_upper-mu1)/sigm1
print('Z-score  of breaking strength 5 is %0.3f' %(z3_low))
print('Z-score of breaking strength 5.5 is %0.3f' %(z3_upper))

p3 = norm.cdf(x3_upper, loc=mu1, scale=sigm1) - norm.cdf(x3_lower, loc=mu1, scale=sigm1) #using cummulative distributive fucntion to calculate probability.
print("The proportion of the gunny bags have a breaking strength between 5 kg/sq. cm and 5.5 kg/ sq cm is", str(round(p3,4)*100)+"%")
```

**Visualisation:**


```python
#Generate x and y for normal distribution curve.
x = np.linspace(mu1-4*sigm1, mu1+4*sigm1, 1000)
y = norm.pdf(x, mu1, sigm1 )
#plot figure size.
plt.figure(figsize=(8,6))
#plot normal distribution curve.
plt.plot(x,y,label='Normal Distribution')
#shade area the curve between 5 < X < 5.5 .
plt.fill_between(x,y, where=((x>=x3_lower) & (x<=x3_upper)), color='orange',alpha=0.5)
#Draws vertical dashed line at the bounds i.e. x3_upper and x3_lower to mark threshold.
plt.axvline(x3_lower, color='red', linestyle='--') # avoid label here , else duplicate entry.
plt.axvline(x3_upper, color='red', linestyle='--', label=f'X in [{x3_lower},{x3_upper}]')
#plot title
plt.title(f'2.3: P({x3_lower} < X < {x3_upper}) = {p3:.4f}')
# x-axis label
plt.xlabel('Breaking Strength(Kg/sq.cm)')# x-axis label
plt.ylabel('Probability Density') #y-axis label.
plt.legend()
plt.grid(True)
plt.show()
```

**Insights:**

*  Only 13% of the gunny bags fall within the strength range of (5-5.5)kg/sq. cm just above the mean.
*  The narrow band around the mean shows the peak of the normal distribution, and highlighting that bags are widely spread across different strength.

**Recommendations:**

* Consider strict process control to reduce this high variability and center more bags in the optimal strength range.



##**2.4 What proportion of the gunny bags have a breaking strength NOT between 3 and 7.5 kg per sq cm.?**

**Calculations:**


```python
x4_lower = 3
x4_upper = 7.5
z4_low = (x4_lower-mu1)/sigm1
z4_upper = (x4_upper-mu1)/sigm1
print('Z-score  of breaking strength 3 is %0.2f' %(z4_low))
print('Z-score of breaking strength 7.5 is %0.2f' %(z4_upper))

p4 = 1- (norm.cdf(x4_upper, loc=mu1, scale=sigm1) - norm.cdf(x4_lower, loc=mu1, scale=sigm1)) #using cummulative distributive fucntion to calculate probability.
print("The proportion of the gunny bags have a breaking strength NOT between 3 kg/sq. cm and 7.5 kg/ sq cm is", str(round(p4,3)*100)+"%")
```

**Visualisation:**

```python
#Generate x and y for normal distribution curve.
x = np.linspace(mu1-4*sigm1, mu1+4*sigm1, 1000)
y = norm.pdf(x, mu1, sigm1 )
#plot figure size.
plt.figure(figsize=(8,6))
#plot normal distribution curve.
plt.plot(x,y,label='Normal Distribution')
#shade regions left tail (x<3) and right tail (x>7.5).
plt.fill_between(x,y, where=((x<x4_lower) | (x>x4_upper)), color='red',alpha=0.5)
#Draws vertical dashed line between x3_upper and x3 lower to mark threshold.
plt.axvline(x4_lower, color='blue', linestyle='--')
plt.axvline(x4_upper, color='blue', linestyle='--', label=f'X not in [{x4_lower},{x4_upper}]')
#plot title
plt.title(f'2.4: P(X ≤ {x4_lower} or X ≥ {x4_upper}) =  {p4:.4f}')
# x-axis label
plt.xlabel('Breaking Strength(Kg/sq.cm)')# x-axis label
plt.ylabel('Probability Density') #y-axis label.
plt.legend()
plt.grid(True)
plt.show()
```

**Insights:**

*  Approximately 13.9% of the gunny bags fall outside the expected performance range, with some weak bags(<3kg/sq. cm) and  others are too strong(>7.5 kg/sq. cm). These are outliers , either too weak or too strong.
*  This suggests weaker bags might break easily during transport, leading to product loss or damage.
* The extra strong bags, although not too risky , use more material than necessary which an increase production costs.

##**3.1 Zingaro has reason to believe that the unpolished stones may not be suitable for printing. Do you think Zingaro is justified in thinking so?**



###**Step 1: Define the Null and Alternate hypothesis**

* Ha (Alternate Hypothesis) is the mean (𝜇) BHI of unpolished stones < 150





### **Step 2 : Select appropriate test.**

* In one-sample t-test, we compare the sample mean with to a known benchmark.

* So we choose 1-sample, one-tailed t-test as we are testing if the mean BHI of unpolished is less than 150, which is required for suitable printing.




### **Step 3: Set significance level $\alpha.$**

* We select $\alpha = 0.05$ based on business context.

### **Step 4: Normality test using Shapiro-Wilk's test & CTL.**

```python
shapiro_test_unpolished=stats.shapiro(df1['Unpolished'])
shapiro_test_treated_polished=stats.shapiro(df1['Treated and Polished'])
print("Sharipo test p-val for Unpolished:")
print(shapiro_test_unpolished)
print("\nSharipo test p-val for Polished:")
print(shapiro_test_treated_polished)
```

**Observations:**

* We can observe Shapiro's stat value for Polished = 0.981 and for unpolished = 0.987.
* We can observe Shapiro's test p-value for Polished = 0.33 and for unpolished = 0.66.
* P-value is much higher than $\alpha $ in this case, thus fail to reject the null hypothesis of normality.
* We can say the data is normally distributed despite visual interpretation showed slight skewness in unpolished data.
* Also, our sample size > 30, i.e. (n = 75 ) which satisfies the Central Limit Theorem which further suggest normal distribution.
* Statistical test > visual interpretation.
* Thus, we can perform t-test as  the combined evidence from Shapiro and CTL provides strong justification for this test.

**T-test Assumptions (Validated):**

Here we did  1-sample, one tailed, t-test.

1. **Continuous variable:** The BHI values are continuous numeric data.
2. **Random sampling:** The sample of unpolished stones is assumed to be randomly selected.
3. **Normality:** The distribution was inspected visually via boxplot and histogram, also validated formally using Shapiro-Wilk test. Though slight skewed, the sample size (n=75) allows robustness of t-test due to CTL.
4. **Data follows t-distribution:** The population standard deviation is unknown, hence the sample follows a t-distribution.



### **Step 5: Collect Data & Calculate p-value**

```python
# Extract all unpolished stones data
unpolished1 = melted_df1[melted_df1['Stone Type'] == 'Unpolished']['BHI']
# Perform 1-sample 1 test using ttest_1samp()
t_stat1, p_value1=stats.ttest_1samp(unpolished1,150,alternative='less')

# Print results.
print(f'T-statistic: {t_stat1}')
print(f'P-value: {p_value1}')
```

###**Step 6: Compare p value with $\alpha$**

```python
# print the conclusion based on p-value
if p_value1 < 0.05:
    print(f'Since the p-value {p_value1} is less than the level of significance, we reject the null hypothesis.')
else:
    print(f'Since the p-value {p_value1} is greater than the level of significance, we fail to reject the null hypothesis.')
```

**Insights:**

* Since the p-value (4.171286997419652e-05) < 0.05, we reject the null hypothesis.
* Negative t-statistic value (-4.16) indicates our sample is below 150.




**The test statistic follows a t-distribution.**

* Let's plot the distribution of the t-test (1-sample, 1-tailed) statistic and check where the computed test statistic lies in the plot.

```python
#Calculate the sample size, sample mean and sample std.
n = len(unpolished1)

# Create 400 points from -4 to 4 to plot t-distribution.
x_t = np.linspace(-4, 4, 400)

#plot figure

plt.figure(figsize=(10, 5))
#df is degree of freedom used to plot curve
plt.plot(x_t, stats.t.pdf(x_t, df=n-1), label='t-distribution')
#Draw vertical line where t-stat lies
plt.axvline(x=t_stat1, color='red', linestyle='--', label=f't-stat = {t_stat1:.2f}')
#Draw green dashed line where t critical value for 5% level (left tailed test)
plt.axvline(x=stats.t.ppf(0.05, df=n-1), color='green', linestyle='--', label='Critical t-value (one-tailed, 5%)')
plt.title('Distribution of t-statistic')
plt.xlabel('t-statistic')
plt.ylabel('Density')
plt.legend()
plt.grid()
plt.show()
```

**Observations:**

* In the plot, the red line (t-stats) lies left of the green (critical value) line.
* This means we reject the null hypothesis at a 5% significance level.

* Therefore, Zingaro is justified in believing unpolished stones are not suitable.

##**3.2 Is the mean hardness of the polished and unpolished stones the same?**

###**Step 1: Define the Null and Alternate hypothesis**

$H_0$: (
𝜇_polished
=
𝜇_unpolished
 )

$H_a$: (
𝜇_polished
≠
𝜇_unpolished
)

-----------------------------------------------------------------------------

* Here $H_0$ (Null Hypothesis) is
mean hardness of the polished (𝜇_polished) = mean hardness of the Unpolished (𝜇_unpolished)

* Here $H_a$ (Alternate Hypothesis) is
mean hardness of the polished (𝜇_polished) ≠ mean hardness of the Unpolished (𝜇_unpolished)

### **Step 2 : Select appropriate test.**

* We choose independent 2-sample, two-tailed  Welch's t-test because we are comparing two independent groups: polished and unpolished stones.
* Our aim is to test whether there is  any difference (smaller or greater) in their average hardness.
* Welch's test is used here because the variance of the two groups are unequal (std unpolished=33.04 and std of polished=15.59).
* Hence, set equal_var = False in the test.

### **Step 3: Set significance level $\alpha.$**

* We select $\alpha = 0.05$ based on business context.

### **Step 4: Normality test using Shapiro-Wilk test & CTL.**

* We already tested normality in above problem using Shapiro & CTL tests and also checked equality of variance by statistical value ( std polished = 15.59, and unpolished 33.04), indicating data is normla but there is no equality of variance.
* We conclude that distribution is normal and we can proceed with 2-sample, Welch's t-test.

**T-test Assumptions (Validated):**

Here we did  independent 2-sample, two tailed, Welch's t-test.

1. **Continuous variable:** The BHI values are continuous numeric in both groups.
2. **Independent groups:** The two samples (unpolished  and polished) stones are assumed to be independent of each other.
3. **Normality in both groups:** The distribution was inspected visually via boxplot and histogram, also validated formally using Shapiro-Wilk test. Normality is satisfied, sample size (n=75) ensures robustness.
4. **Unequal Variance allowed:** Variances are unequal, hence Welch's correction applied (no assumption of equal variance is required).

Therefore, our T-assumptions are validated

### **Step 5: Collect data & Calculate p-value**

```python
# Extract all unpolished and unpolished stones data
unpolished1 = melted_df1[melted_df1['Stone Type'] == 'Unpolished']['BHI']
polished1= melted_df1[melted_df1['Stone Type'] == 'Treated and Polished']['BHI']

#Perform 2-sample t-test using ttest_ind()
# std of unpolished=33.04 and std of polished=15.59.
#So, unequal variance,keep equal_var=False. Welch's t-test.
t_stat2, p_value2=stats.ttest_ind(polished1,unpolished1,equal_var=False,alternative='two-sided')

#Print results
print(f'T-statistic: {t_stat2}')
print(f'P-value: {p_value2}')
```

###**Step 6: Compare p value with $\alpha$**

```python
# print the conclusion based on p-value
if p_value2 < 0.05:
    print(f'Since the p-value {p_value2} is less than the level of significance, we reject the null hypothesis.')
else:
    print(f'Since the p-value {p_value2} is greater than the level of significance, we fail to reject the null hypothesis.')
```

**Insights:**

* T-stats value is positive (3.24) suggests that the average hardness of the polished stones is higher than that of unpolished stones.
* P-value (0.00158) < 0.05 , we reject the null hypothesis.

**The test statistic follows a Welch's t-distribution.**

* Let's plot the distribution of the Welch's t-statistic (2-sample, 2-tailed) and check where the computed test statistic lies in the plot.

```python
#Calculate the sample size, sample variance.
n1, n2 = len(polished1), len(unpolished1) #sample size
s1,s2 = np.var(polished1,ddof=1), np.var(unpolished1,ddof=1) #sample variances

#Calculate Welch's degrees of freedom
df_welch = (s1/n1 + s2/n2)**2 / ((s1**2)/((n1**2)*(n1-1)) + (s2**2)/((n2**2)*(n2-1)))

# Create 400 points from -4 to 4 to plot t-distribution.
x_w = np.linspace(-4, 4, 300)

#plot figure

plt.figure(figsize=(8, 5))
#plot t-distribution with welch's degrees of freedom
plt.plot(x_w, stats.t.pdf(x_w, df=df_welch), label=f"Welch's DoF (df= {df_welch:.1f})")
#Draw vertical red line at t-stat
plt.axvline(x=t_stat2, color='red', linestyle='--', label=f"Welch's t-stat = {t_stat2:.2f}")
#Draw green dashed line for two-tailed test at 5% level (±2.5%)
plt.axvline(-t.ppf(0.975, df=df_welch), color='green', linestyle='--', label='Critical t-values (±2.5%)')
plt.axvline(t.ppf(0.975, df=df_welch), color='green', linestyle='--')
plt.title("Distribution of Welch's t-statistic")
plt.xlabel('t-statistic')
plt.ylabel('Density')
plt.legend()
plt.grid()
plt.show()
```

**Observations:**

* In the plot, the red dashed lines (computed t-stats) lies outside the green lines, which represent the critical values for a two tailed test at 5% significance level (i.e. 2.5% in each tail).
* Therefore, we reject the null hypothesis at 5% significance level.

### **Step 7: Conclusion**

* Based on the statistical evidence, we reject the null hypothesis at 5% significance level.
* There is sufficient evidence to conclude that there is a significant difference in the mean hardness between polished and unpolished stones.
* Furthermore, the positive t-statistics indicates that polished stones tend to have a higher mean BHI compared to unpolished stones.
* This finding is relevant for Zingaro stone printing, as it shows the need to consider surface type when assessing stone suitability for printing.

###**Outlier Detection and Treatment**





```python
# function to detect outlier in Response based on alloy 1 & 2

def detect_outlier(df,col):
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - (1.5 * IQR)
  upper_bound = Q3 + (1.5 * IQR)
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  outlier_count = len(outliers)
  return outliers, outlier_count

#Alloy1
outliers_alloy1, outlier_count1 = detect_outlier(alloy1_df,'Response')

print(f'Outliers in Alloy 1:\n{outliers_alloy1}')
print(f'\nTotal number of outliers in Alloy 1: {outlier_count1}')

#Alloy2

outliers_alloy2, outlier_count2 = detect_outlier(alloy2_df,'Response')

print(f'\n\nOutliers in Alloy 2:\n{outliers_alloy2}')
print(f'\nTotal number of outliers in Alloy 2: {outlier_count2}')
```

##**Hypothesis Testing**

###**4.1 How does the hardness of implants vary depending on dentists?**

####**Step 1: Define Null and Alternative hypotheses.**

**Alloy 1 Hypotheses:**

The null and alternative hypotheses  for Alloy 1 can be given as:

> $H_0$: The mean hardness of implants is equal across all dentists for alloy 1.<br>

> $H_a$: The mean hardness of implants differs in at least one dentists for alloy 1.





**Alloy 2 Hypotheses:**

The null and alternative hypotheses  for Alloy 2 can be given as:

> $H_0$: The mean hardness of implants is equal across all dentists for alloy 2.<br>

> $H_a$: The mean hardness of implants differs in at least one dentists for alloy 2.

####**Step 2: Select the Appropriate test.**

This problem needs us to find how mean hardness(Response) vary across all dentists. **One-way ANOVA is an appropriate test here for Alloy 1 & 2**,provided normality and equality of variance assumptions are verified.

**One-way ANOVA Test Assumptions for Alloy 1 & 2:**
The results of this test is reliable if the assumptions of normality and equality are satisfied.

1.**Normality test**:
**Shapiro-Wilk's** test is applied to the response variable.

* We will test the null hypothesis and Alternative Hypothesis.


> $H_0$:  The mean hardness of implant (Response) follow a normal distribution for Alloy . <br>

> $H_𝑎$: The mean hardness (Response) do not follow a normal distribution.

2.**Equality of variance test**:
**Levene's test** is applied to the response variable.

* We will test the null hypothesis and Alternative Hypothesis.


> $H_0$: The hardness(Response) for all dentist variances are equal against the alternative hypothesis.<br>

> $H_𝑎$: At least one variance is different from the rest.

**Shapiro's Test for both Alloy 1.**

```python
#Alloy 1
print("Shapiro-Wilk Normality Test for Alloy 1:\n")
normality_result1 = {} # created dict to store normality p-values.
#loop through each dentist group
for dentist in alloy1_df['Dentist'].unique():
  # Extract hardness(Response) values for this dentist group
  group_data = alloy1_df[alloy1_df['Dentist'] == dentist]['Response']
  # Perform Shapiro-Wilk test for normality
  statistic, p_val = stats.shapiro(group_data)
  # Store the p-value in the dictionary
  normality_result1[dentist] = p_val
  print(f'Dentist {dentist}: p-value = {p_val:.4f}')

#Interpretation of normality results
print('\nInterpretation of normality results:')
#loop through each p-value
for dentist, p in normality_result1.items():
  print(f"Dentist {dentist}:{'Normality distributed for Alloy 1' if p > 0.05 else 'Not normally distributed for Alloy 1'}")


```

**Insight:**

* p-value for all dentists based on alloy 1 is above 5% significance level.
* Therefore, the hardness (Response) follows a normal distribution, we fail to reject the null hypothesis.


**Shapiro's Test for both Alloy 2.**

```python
#Alloy 2
print("Shapiro-Wilk Normality Test for Alloy 2")
normality_result2 = {} # created dict to store normality p-values.
#loop through each dentist group
for dentist in alloy2_df['Dentist'].unique():
  # Extract hardness(Response) values for this dentist group
  group_data = alloy2_df[alloy2_df['Dentist'] == dentist]['Response']
  # Perform Shapiro-Wilk test for normality
  statistic, p_val = stats.shapiro(group_data)
  # Store the p-value in the dictionary
  normality_result2[dentist] = p_val
  print(f'Dentist {dentist}: p-value = {p_val:.4f}')

#Interpretation of normality results
print('\nInterpretation of normality results:')
#loop through each p-value
for dentist, p in normality_result2.items():
  print(f"Dentist {dentist}:{'Normality distributed for Alloy 2' if p > 0.05 else 'Not normally distributed for Alloy 2'}")
```

**Insight:**

* p-value for dentists (1,2,3 & 5) based on alloy 2 is above 5% significance level, indicating normality.
* So, we fail to reject the null hypothesis of normality.
* Therefore, we have sufficient evidence that the hardness (Response) follows a normal distribution for Dentists (1,2,3 & 5).
* However, p-value for dentists 4 based on alloy 2 is less than 5% significance level, indicating data is not normal.
* Since, the normality of assumption is violated for at least one group (Dentist 4).

This problem needs us to find how mean hardness(Response) vary across all methods.**One-way ANOVA is an appropriate test for Alloy 1 & 2** here, provided normality and equality of variance assumptions are verified.

**One-way ANOVA Test Assumptions for Alloy 1 & Alloy 2:**
The results of this test is reliable if the assumptions of normality and equality are satisfied.

1.**Normility test**:

**Shapiro-Wilk's** test is applied to the response variable.

* We will test the null hypothesis and Alternative Hypothesis.

> $H_0$:  The mean hardness of implant (Response) follow a normal distribution against the alternative hypothesis. <br>

> $H_𝑎$: The mean hardness (Response) do not not follow a normal distribution.

2.**Equality of variance test**:
**Levene's test** is applied to the response variable.

* We will test the null hypothesis and Alternative Hypothesis.

> $H_0$: The hardness(Response) for all method variances are equal against the alternative hypothesis.<br>

> $H_𝑎$: At least one variance is different from the rest.

**Shapiro's Test for Alloy 1.**

```python
#Alloy 1
print("Shapiro-Wilk Normality Test for Alloy 1:\n")
normality_result3 = {} # created dict to store normality p-values.
#loop through each method group
for methods in alloy1_df['Method'].unique():
  # Extract hardness(Response) values for this method group
  group_data1 = alloy1_df[alloy1_df['Method'] == methods]['Response']
  # Perform Shapiro-Wilk test for normality
  statistic, p_val = stats.shapiro(group_data1)
  # Store the p-value in the dictionary
  normality_result3[methods] = p_val
  print(f'Method {methods}: p-value = {p_val:.4f}')

#Interpretation of normality results
print('\nInterpretation of normality results:')
#loop through each p-value
for methods, p in normality_result3.items():
  print(f"Method {methods}:{'Normality distributed for Alloy 1' if p > 0.05 else 'Not normally distributed for Alloy 1'}")
```

```python
#ALLOY 1
# Create groups for Levene's test for all combinations of Dentist and Method.
groups_alloy1 = [] #empty list
groups_labels_alloy1 = [] #empty list
#loop through each combination of dentist and method for Alloy 1
for d in alloy1_df['Dentist'].unique():
    for m in alloy1_df['Method'].unique():
        # Extract hardness(Response) values for this combination
        group_data = alloy1_df[(alloy1_df['Dentist'] == d) & (alloy1_df['Method'] == m)]['Response']
        if len(group_data) > 0:  #include only non-empty groups
            groups_alloy1.append(group_data.values)
            groups_labels_alloy1.append(f'Dentist {d} - Method {m}')

# Perform Levene's test
if len(groups_alloy1) > 1:
    lev_stat_alloy1, lev_p_alloy1 = stats.levene(*groups_alloy1) # * to unpack list
    print("Levene's test for alloy 1:")
    print(f'Levene Statistic: {lev_stat_alloy1:.4f}')
    print(f'p-value: {lev_p_alloy1:.4f}')

    print(f"Conclusion: {'Fail to reject H₀' if lev_p_alloy1 > 0.05 else 'Reject H₀'}")

    if lev_p_alloy1 > 0.05:
        print("Equal variances across all dentist-method combinations.")
    else:
        print("At least one variance is different across all dentist-method combinations.")
```

**Insights:**

* P-value for Alloy 1 is above 5 % significance level.
* So, we fail to reject the null hypothesis.
* Therefore, equal variances across all dentist-method combinations, and it satisfies our assumption.

**Post-Adhoc analysis for Alloy 1 (significant interaction)**

* If the resultant p-value leads to rejection of null hypothesis in an ANOVA, then we perform Post-Adhoc analysis, especially when we have significant main effect interaction with two or more levels.

* Since, our null hypothesis is rejected for Alloy 1, we will perform post-adhoc analysis.

* ANOVA tells us there is significant overall effect of the interaction but does not tell where the difference lie.

* The interaction term indicates that the effect of one factor depends on the level of the other factor, but it doesn't tell us which specific combination of factors levels are significantly different from each other.

* This can be identified using Post-Adhoc test as it allows us to do pairwise comparison between combinations of factor levels to determine which specific combinations (Dentist-Method) significantly different from each other or significantly results in different hardness value.

**Tukey's HSD pairwise comparison for Alloy 1**

```python
# Alloy 1

#Create interactionn factor (Dentist-Method)
alloy1_df_adhoc = alloy1_df.copy() #create a copy of subset data
# Ensure both Dentist and Method columns are treated as strings before concatenating
alloy1_df_adhoc['Dentist-Method'] = alloy1_df_adhoc['Dentist'].astype(str) + '-' + alloy1_df_adhoc['Method'].astype(str)

#perform Tukey HSD test for pairwise comparisons
print("Tukey's HSD pairwise comparison for Alloy 1:")
tukey_res_alloy1 = pairwise_tukeyhsd(endog=alloy1_df_adhoc['Response'], groups=alloy1_df_adhoc['Dentist-Method'], alpha=0.05)
#endog - specifies dependent variable
print(tukey_res_alloy1)


#Calculate and display group means of Response for each combination of dentist-method
print("\n\nGroup Means (Dentist x Method) for Alloy 1:")
group_means_alloy1 = alloy1_df.groupby(['Dentist', 'Method'],observed=True)['Response'].agg(['mean', 'std', 'count'])
print(group_means_alloy1)


#Identify significantly different combinations
print("\n\nSignificantly Different Combinations (Dentist x Method) for Alloy 1:")
#This converts the Tukey HSD results into Df for easier analysis and filtering.
tukey_df_alloy1 = pd.DataFrame(data = tukey_res_alloy1.summary().data[1:],columns = tukey_res_alloy1.summary().data[0])
print(tukey_df_alloy1)

#Convert p-adj column to numeric for filtering
#This filters DF to identify only pairs of the groups that have significant diffrences (p-adj < 0.05)
tukey_df_alloy1['p-adj'] = pd.to_numeric(tukey_df_alloy1['p-adj'])
significant_combinations_alloy1 = tukey_df_alloy1[tukey_df_alloy1['p-adj'] < 0.05]


print(f"\n\nSignificant Pairwise Differences for Alloy 1 (p < 0.05):")
if len(significant_combinations_alloy1) > 0:
  for idx, row in significant_combinations_alloy1.iterrows():
#This prints the pairs of groups that have a significant difference, along with their adjusted p-values.
       print(f" {row['group1']} vs {row['group2']}:p-adj = {row['p-adj']:.4f}")
else:
    print("No significant pairwise differences found.")
```
